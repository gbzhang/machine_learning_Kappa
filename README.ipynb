# Project Description

This project is completed using Jupyter Notebook, including steps such as data feature extension, data exploration, model training and evaluation, and model interpretation. All code files, data files, and model files are organized as follows:

## File Structure
    csv/

    df.csv # Complete feature data after Magpie extension

    df_k.csv # Optimized feature data (used for model training)

    new_k.csv # Model prediction results vs actual results comparison

    new_magpie.csv # New data for model generalization validation (already Magpie-extended)

    test.csv # Original data source file

    xgb_new.csv # XGBoost prediction results on new data

    model/

    feature_order.pkl # Feature order (for model input)

    scaler.pkl # Standardizer (for data preprocessing)

    xgb_k_model.pkl # Trained XGBoost model

    png/ # Directory for image files

    magpie.ipynb # Data feature extension and preprocessing

    kappa.ipynb # Data distribution analysis (e.g., Kappa statistics)

    xgb.ipynb # XGBoost model training, prediction, and SHAP analysis

    lgb.ipynb # LightGBM model training and prediction

    rf.ipynb # Random Forest model training and prediction

## Execution Workflow

1. **`magpie.ipynb`**
   - Perform feature expansion and preprocessing on the raw data (`test.csv`)
   - Generate `df.csv` (full feature data after Magpie expansion)
   - Further optimize features to generate `df_k.csv` (used for model training)
   - Simultaneously generate `new_magpie.csv` (used for model generalization validation)

2. **`kappa.ipynb`**
   - Read `df_k.csv`
   - Analyze data distribution
   - Help understand data quality and class distribution

3. **Model Training and Comparison** (run the following three notebooks separately)
   - **`xgb.ipynb`**: Training and prediction using XGBoost
   - **`lgb.ipynb`**: Training and prediction using LightGBM
   - **`rf.ipynb`**: Training and prediction using Random Forest

   All three models are trained based on `df_k.csv`.

4. **Select and Save the XGBoost Model** (completed in `xgb.ipynb`)
   - Save model files to the `model/` directory:
     - `feature_order.pkl`: Feature order
     - `scaler.pkl`: Standardizer
     - `xgb_k_model.pkl`: Trained XGBoost model

   - Generate prediction results:
     - `xgb_new.csv`: Prediction results on new data (`new_magpie.csv`)
     - `new_k.csv`: Comparison between model predictions and actual results

5. **`shap.ipynb`**
   - Load the saved XGBoost model (`xgb_k_model.pkl`) and related preprocessing files
   - Perform SHAP analysis on the model to interpret feature importance and prediction behavior

## Notes

- All data files (`.csv`) are stored in the `csv/` folder, and model files are stored in the `model/` folder.
- The raw data source is `test.csv`; all model training and predictions are based on this file or its derived files.
- The final selected model is **XGBoost**, and model interpretation (SHAP) has been performed.
