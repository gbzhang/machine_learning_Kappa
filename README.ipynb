# 项目说明

本项目基于 Jupyter Notebook 完成，包含数据特征拓展、数据探索、模型训练与评估、模型解释等步骤。所有代码文件、数据文件及模型文件均组织如下：

## 文件结构
    csv/

    df.csv # 使用 Magpie 拓展后的完整特征数据

    df_k.csv # 优化特征后的数据（用于模型训练）

    new_k.csv # 模型预测结果与实际结果对比

    new_magpie.csv # 用于模型泛化能力验证的新数据（已做 Magpie 拓展）

    test.csv # 原始数据源文件

    xgb_new.csv # 使用 XGBoost 对新数据的预测结果

    model/

    feature_order.pkl # 特征顺序（用于模型输入）

    scaler.pkl # 标准化器（用于数据预处理）

    xgb_k_model.pkl # 训练好的 XGBoost 模型

    png/ # 存放图像文件

    magpie.ipynb # 数据特征拓展与预处理

    kappa.ipynb # 查看数据分布（如 Kappa 统计）

    xgb.ipynb # XGBoost 模型训练与预测

    lgb.ipynb # LightGBM 模型训练与预测

    rf.ipynb # 随机森林模型训练与预测

    shap.ipynb # 对 XGBoost 模型进行 SHAP 分析


## 执行顺序说明

1. **`magpie.ipynb`**
   - 对原始数据（`test.csv`）进行特征拓展与预处理
   - 生成 `df.csv`（Magpie 拓展后的全特征数据）
   - 进一步优化特征，生成 `df_k.csv`（用于模型训练）
   - 同时生成 `new_magpie.csv`（用于模型泛化能力验证）

2. **`kappa.ipynb`**
   - 读取 `df_k.csv`
   - 分析数据分布
   - 帮助理解数据质量与类别分布情况

3. **模型训练与比较**（分别运行以下三个笔记本）
   - **`xgb.ipynb`**：使用 XGBoost 进行训练与预测
   - **`lgb.ipynb`**：使用 LightGBM 进行训练与预测
   - **`rf.ipynb`**：使用随机森林进行训练与预测

   这三个模型均基于 `df_k.csv` 进行训练。

4. **选定 XGBoost 模型并保存**（在 `xgb.ipynb` 中完成）
   - 保存模型文件至 `model/` 目录：
     - `feature_order.pkl`：特征顺序
     - `scaler.pkl`：标准化器
     - `xgb_k_model.pkl`：训练好的 XGBoost 模型

   - 生成预测结果：
     - `xgb_new.csv`：对新数据（ `new_magpie.csv`）的预测结果
     - `new_k.csv`：模型预测结果与实际结果对比

5. **`shap.ipynb`**
   - 加载已保存的 XGBoost 模型（`xgb_k_model.pkl`）及相关预处理文件
   - 对模型进行 SHAP 分析，解释特征重要性及预测行为

## 注意事项

- 所有数据文件（`.csv`）均存放在 `csv/` 文件夹下，模型文件存放在 `model/` 文件夹下。
- 原始数据源为 `test.csv`，所有模型训练与预测均基于该文件或其衍生文件。
- 模型最终选定为 **XGBoost**，并进行了模型解释（SHAP）。
